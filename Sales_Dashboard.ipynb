```sql
SELECT
    ProductName,
    ROUND(SUM(SalesAmount), 2) AS Sales
FROM gold_sales
WHERE OrderStatus = 'Completed'
GROUP BY ProductName
ORDER BY Sales DESC
LIMIT 10;
```

```sql
SELECT
    SalesChannel,
    ROUND(SUM(SalesAmount), 2) AS Sales
FROM gold_sales
WHERE OrderStatus = 'Completed'
GROUP BY SalesChannel;
```

```sql
SELECT
    ProductCategory,
    ROUND(SUM(SalesAmount), 2) AS Sales
FROM gold_sales
WHERE OrderStatus = 'Completed'
GROUP BY ProductCategory
ORDER BY Sales DESC;
```

```sql
SELECT
    DATE_TRUNC('month', TransactionDate) AS Month,
    ROUND(SUM(SalesAmount), 2) AS Sales
FROM gold_sales
WHERE OrderStatus = 'Completed'
GROUP BY DATE_TRUNC('month', TransactionDate)
ORDER BY Month;
```

```sql
SELECT
    ROUND(
        SUM(SalesAmount) / COUNT(DISTINCT TransactionID),
        2
    ) AS Average_Order_Value
FROM gold_sales
WHERE OrderStatus = 'Completed';
```

```sql
SELECT
    SUM(Quantity) AS Units_Sold
FROM gold_sales
WHERE OrderStatus = 'Completed';
```

```sql
SELECT
    COUNT(DISTINCT TransactionID) AS Total_Orders
FROM gold_sales
WHERE OrderStatus = 'Completed';
```

```sql
SELECT
    ROUND(SUM(SalesAmount), 2) AS Total_Sales
FROM gold_sales
WHERE OrderStatus = 'Completed';
```
